# Código-fonte RPA (Pyhton)

````python
from bs4 import BeautifulSoup
import requests

# 1) Simulando o HTML que viria de uma página real
html = """
<!doctype html>
<html lang="pt-br">
<head>
  <meta charset="utf-8" />
  <title>Clientes</title>
</head>
<body>
  <table id="clientes">
    <thead>
      <tr>
        <th>Nome</th>
        <th>Email</th>
        <th>Saldo</th>
        <th>Perfil</th>
      </tr>
    </thead>
    <tbody>
      <tr>
        <td class="nome">Ana Silva</td>
        <td class="email">ana@email.com</td>
        <td class="saldo">R$ 12.500,00</td>
        <td class="perfil">Conservador</td>
      </tr>
      <tr>
        <td class="nome">Bruno Lima</td>
        <td class="email">bruno@email.com</td>
        <td class="saldo">R$ 3.200,00</td>
        <td class="perfil">Moderado</td>
      </tr>
      <tr>
        <td class="nome">Carla Souza</td>
        <td class="email">carla@email.com</td>
        <td class="saldo">R$ 28.900,00</td>
        <td class="perfil">Arrojado</td>
      </tr>
    </tbody>
  </table>
</body>
</html>
"""

# Se você tivesse um site real, usaria isso no lugar da variável acima:
# URL_PAGINA = "https://www.site-real-com-a-tabela.com.br"
# html = requests.get(URL_PAGINA).text

# 2) Ler a tabela de clientes
soup = BeautifulSoup(html, "html.parser")
linhas = soup.select("#clientes tbody tr")

clientes = []

for linha in linhas:
    colunas = linha.find_all("td")

    if len(colunas) >= 4: # Garantir que a linha tem as colunas necessárias
        cliente = {
            "nome":  colunas[0].get_text(strip=True),
            "email": colunas[1].get_text(strip=True),
            "saldo": colunas[2].get_text(strip=True),
            "perfil": colunas[3].get_text(strip=True),
        }
        clientes.append(cliente)

# 3) Mostrar o resultado no Colab
print("Clientes encontrados:")
for cliente in clientes:
    print(cliente)

# 4) Enviar para o n8n
N8N_WEBHOOK = "https://dion8nsantander.app.n8n.cloud/webhook-test/Clientes"

print("\nEnviando para o n8n...")
try:
    resposta = requests.post(N8N_WEBHOOK, json={"clientes": clientes})
    print(f"Status do envio: {resposta.status_code}")
except Exception as e:
    print(f"Erro ao enviar para o n8n: {e}")
````
